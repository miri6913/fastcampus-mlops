## **Chapter 2. 머신러닝 상품화를 위한 기술**

### Why do we need data warehouse where we can store it in database?

- Database is for fast, live, moment-to-moment operations, like updating a customer's order.
- Data warehouse is for storing clean, structured data specifically to run big analytical queries and reports.
- Data lake is for dumping raw, unstructured, or mixed data cheaply, to be structured and analyzed later.

### Inference in easy language

- Inference is the act of a trained model taking new input and producing a prediction or output in real time.
- Inference is a big deal because it's where the model actually delivers value in real time, and it forces tradeoffs between speed, cost, and accuracy that training doesn't.

## **Chapter 3. 머신러닝 상품화를 위한 인프라와 도구들**

### Difference between container and virtual machine and when to use each?

- A container packages an app and its dependencies while sharing the host's operating system kernel, making it lightweight and fast to start.
- A virtual machine emulates an entire separate computer with its own operating system, making it heavier but fully isolated.
- Use containers for quickly packaging and shipping apps consistently, and use virtual machines when you need strong isolation between different operating systems or untrusted workloads.
