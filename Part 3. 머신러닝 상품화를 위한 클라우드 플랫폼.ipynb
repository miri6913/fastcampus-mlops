### MLFlow and Airflow, when to use each

- MLflow is for tracking your machine learning experiments, things like which parameters you used, what metrics you got, and saving your models.
- Airflow is for scheduling and orchestrating workflows in general, like running a sequence of tasks in order, on a schedule, such as every night at midnight.
- Use MLflow when you need to track and compare experiments or manage model versions, and use Airflow when you need to schedule and automate a pipeline of tasks running in sequence.

### Loss, and accuracy?

- **Accuracy:** the percentage of predictions the model got right.
- **Loss:** a penalty score for how far the model's predictions are from the correct answers, and training tries to make it smaller.
- For the penalty, a good prediction (0.99 on the correct answer) gets a **small** penalty, and a bad one (0.01 on the correct answer) gets a **large** penalty.
